```python
import os
import subprocess

# --- 1. 安装 Java 17 (OpenJDK) ---
print("📦 Installing Java 17 via Conda...")
subprocess.run("wget -q https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh -O miniconda.sh", shell=True)
subprocess.run("bash miniconda.sh -b -p /usr/local/miniconda3", shell=True)
os.environ['PATH'] = "/usr/local/miniconda3/bin:" + os.environ['PATH']
subprocess.run("conda install -y -c conda-forge openjdk=17", shell=True, stdout=subprocess.DEVNULL)

```

```python
# --- 2. 安装 uv ---
print("⚡ Installing uv (Fast Python package manager)...")
subprocess.run("curl -LsSf https://astral.sh/uv/install.sh | sh", shell=True)
os.environ['PATH'] = f"/root/.cargo/bin:{os.environ['PATH']}"
```

```python
# --- 3. 创建虚拟环境并安装 PyTorch (GPU Version) ---
# 使用 uv 创建 venv，并强制安装 CUDA 12.4 版本的 PyTorch
print("🏗️ Creating virtual environment with PyTorch (CUDA 12.4)...")
subprocess.run("uv venv .venv --seed", shell=True)
# 这一行是关键：确保 PyTorch 能找到 Colab 的 GPU
subprocess.run("""
source .venv/bin/activate && \
uv pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu124
""", shell=True)
```

```python
# --- 4. 验证 GPU ---
print("\n✅ Verifying GPU availability:")
subprocess.run("source .venv/bin/activate && python -c 'import torch; print(f\"CUDA Available: {torch.cuda.is_available()}\"); print(f\"Device: {torch.cuda.get_device_name(0)}\")'", shell=True)

print("\n🎉 Setup complete! Please **Restart Runtime** (Runtime -> Restart Runtime) before running Cell 2.")
```

```python
from google.colab import files
import os

# 1. 上传文件
uploaded = files.upload()
if not uploaded:
    print("No file uploaded.")
    raise SystemExit

# 2. 获取文件名
pdf_name = list(uploaded.keys())[0]
base_name = os.path.splitext(pdf_name)[0]

print(f"Processing: {pdf_name}")
```

```python
!source .venv/bin/activate
```

```python
!opendataloader-pdf "{pdf_name}" -o ./output -f markdown,json --keep-line-breaks
```

```python
output_md = f"./output/{base_name}.md"
if os.path.exists(output_md):
    print("\n" + "="*30 + " MARKDOWN OUTPUT " + "="*30)
    with open(output_md, 'r', encoding='utf-8') as f:
        content = f.read()
        print(content[:2000])  # 打印前2000字符
        if len(content) > 2000:
            print("\n... (truncated)")
else:
    print("❌ Conversion failed or file not found.")
```

```python
# @title 📥 4. 下载转换结果

import shutil
from google.colab import files

# 打包输出文件夹
archive_name = "converted_pdfs"
shutil.make_archive(archive_name, 'zip', './output')

# 触发下载
files.download(f'{archive_name}.zip')
print("✅ Download started!")
```
